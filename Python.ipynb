```python
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_csv("/kaggle/input/datasets/valakhorasani/gym-members-exercise-dataset/gym_members_exercise_tracking.csv")

data["BMI"] = data["Weight (kg)"] / (data["Height (m)"] ** 2)
data["Performance"] = data["Calories_Burned"] / data["Session_Duration (hours)"]

# 1. BMI Distribution
plt.figure(figsize=(8,5))
plt.hist(data["BMI"], bins=12, color="skyblue")
plt.xlabel("BMI")
plt.ylabel("Population")
plt.title("BMI Distribution")
plt.show()

# 2. BMI vs Calories Burned
plt.figure(figsize=(8,5))
plt.scatter(data["BMI"], data["Calories_Burned"], alpha=0.6, color="red")
plt.xlabel("BMI")
plt.ylabel("Calories Burned")
plt.title("BMI vs Calories Burned")
plt.show()

# 3. Age vs Calories Burned
plt.figure(figsize=(8,5))
plt.scatter(data["Age"], data["Calories_Burned"], alpha=0.6, color="green")
plt.xlabel("Age")
plt.ylabel("Calories Burned")
plt.title("Age vs Calories Burned")
plt.show()

# 4. Workout Duration vs Calories Burned
plt.figure(figsize=(8,5))
plt.scatter(data["Session_Duration (hours)"], data["Calories_Burned"], alpha=0.6, color="orange")
plt.xlabel("Workout Duration (hours)")
plt.ylabel("Calories Burned")
plt.title("Workout Duration vs Calories Burned")
plt.show()

# 5. Heart Rate vs Calories Burned
plt.figure(figsize=(8,5))
plt.scatter(data["Avg_BPM"], data["Calories_Burned"], alpha=0.6, color="purple")
plt.xlabel("Average BPM")
plt.ylabel("Calories Burned")
plt.title("Heart Rate vs Calories Burned")
plt.show()

# 6. Fat % vs Calories Burned
plt.figure(figsize=(8,5))
plt.scatter(data["Fat_Percentage"], data["Calories_Burned"], alpha=0.6, color="brown")
plt.xlabel("Fat Percentage")
plt.ylabel("Calories Burned")
plt.title("Fat % vs Calories Burned")
plt.show()

# 7. Workout Type vs Calories Burned
plt.figure(figsize=(8,5))
data.groupby("Workout_Type")["Calories_Burned"].mean().plot(kind="bar", color="pink")
plt.xlabel("Workout Type")
plt.ylabel("Average Calories Burned")
plt.title("Calories Burned by Workout Type")
plt.show()

```
